## 빅쿼리 기본 설정

```python
import pandas as pd
from google.cloud import bigquery

client = bigquery.Client()

job_config = bigquery.QueryJobConfig(
    default_dataset="tap-part4.share_office" 
) # 내가 사용하는 프로젝트와 데이터 세트명을 미리 변수로 넣어준다.
```

## 데이터 불러오기

```python
# trial_visit_info
query = "SELECT * FROM trial_visit_info"
trial_visit_info_df = client.query(query, job_config=job_config).to_dataframe()

# trial_register
query = "SELECT * FROM trial_register"
trial_register_df = client.query(query, job_config=job_config).to_dataframe()

# site_area
query = "SELECT * FROM site_area"
site_area_df = client.query(query, job_config=job_config).to_dataframe()

# trial_payment
query = "SELECT * FROM trial_payment"
trial_payment_df = client.query(query, job_config=job_config).to_dataframe()

# trial_access_log
query = "SELECT * FROM trial_access_log"
trial_access_log_df = client.query(query, job_config=job_config).to_dataframe()
```
